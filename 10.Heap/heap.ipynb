````java
import java.util.*;

class Heap {

    private int[] arr; // heap array
    private int maxSize; // max capacity
    private int heapSize; // current size

    // constructor
    Heap(int maxSize) {
        this.maxSize = maxSize;
        this.heapSize = 0;
        this.arr = new int[maxSize];
    }

    // index of parent
    int parent(int i) {
        return (i - 1) / 2;
    }

    // left child
    int lChild(int i) {
        return 2 * i + 1;
    }

    // right child
    int rChild(int i) {
        return 2 * i + 2;
    }

    // get max
    int getMax() {
        return arr[0];
    }

    int curSize() {
        return heapSize;
    }

    // insert key
    void insertKey(int x) {

        if (heapSize == maxSize) {
            System.out.println("Overflow: Could not insertKey");
            return;
        }

        heapSize++;
        int i = heapSize - 1;
        arr[i] = x;

        // fix max-heap property
        while (i != 0 && arr[parent(i)] < arr[i]) {
            swap(i, parent(i));
            i = parent(i);
        }

    }

    // increase key
    void increaseKey(int i, int newVal) {
        arr[i] = newVal;

        while (i != 0 && arr[parent(i)] < arr[i]) {
            swap(i, parent(i));
            i = parent(i);
        }
    }

    // remove max
    int removeMax() {

        if (heapSize <= 0)
            return Integer.MIN_VALUE;

        if (heapSize == 1) {
            heapSize--;
            return arr[0];
        }

        int root = arr[0];
        arr[0] = arr[heapSize - 1];
        heapSize--;

        heapify(0);

        return root;
    }

    // delete key
    void deleteKey(int i) {
        increaseKey(i, Integer.MAX_VALUE);
        removeMax();
    }

    // heapify
    void heapify(int i) {

        int l = lChild(i);
        int r = rChild(i);
        int largest = i;

        if (l < heapSize && arr[l] > arr[largest])
            largest = l;

        if (r < heapSize && arr[r] > arr[largest])
            largest = r;

        if (largest != i) {
            swap(i, largest);
            heapify(largest);
        }
    }

    // print heap
    void print() {
        System.out.print("The heap is :- ");
        for (int i = 0; i < heapSize; i++)
            System.out.print(arr[i] + " ");
        System.out.println();
    }

    // swap utility
    void swap(int i, int j) {
        int temp = arr[i];
        arr[i] = arr[j];
        arr[j] = temp;
    }

}
````

````java
// driver / main
Heap maxheap = new Heap(15);

System.out.println("Entered 6 keys:- 3, 10, 12, 8, 2, 14");

maxheap.insertKey(3);
maxheap.insertKey(10);
maxheap.insertKey(12);
maxheap.insertKey(8);
maxheap.insertKey(2);
maxheap.insertKey(14);

maxheap.print();
System.out.println("The current size of the heap is " + maxheap.curSize());
System.out.println("The current maximum element is " + maxheap.getMax());

maxheap.deleteKey(2);
System.out.println("The current size of the heap is " + maxheap.curSize());
maxheap.print();

maxheap.insertKey(15);
maxheap.print();

maxheap.insertKey(5);
maxheap.print();

System.out.println("The current size of the heap is " + maxheap.curSize());
System.out.println("The current maximum element is " + maxheap.getMax());
````

````java
import java.util.ArrayList;

// max heap
class Heap<T extends Comparable<T>> {

  private ArrayList<T> list;

  public Heap() {
    list = new ArrayList<>();
  }

  private void swap(int first, int second) {
    T temp = list.get(first);
    list.set(first, list.get(second));
    list.set(second, temp);
  }

  private int parent(int index) {
    return (index - 1) / 2;
  }

  private int left(int index) {
    return index * 2 + 1;
  }

  private int right(int index) {
    return index * 2 + 2;
  }

  public void insert(T value) {
    list.add(value);
    upheap(list.size() - 1);
  }

  private void upheap(int index) {
    if (index == 0) {
      return;
    }
    int p = parent(index);
    if (list.get(index).compareTo(list.get(p)) > 0) {
      swap(index, p);
      upheap(p);
    }
  }

  public T remove() throws Exception {
    if (list.isEmpty()) {
      throw new Exception("Removing from an empty heap!");
    }

    T temp = list.get(0);

    T last = list.remove(list.size() - 1);
    if (!list.isEmpty()) {
      list.set(0, last);
      downheap(0);
    }

    return temp;
  }

  private void downheap(int index) {
    int max = index;
    int left = left(index);
    int right = right(index);

    if (left < list.size() && list.get(max).compareTo(list.get(left)) < 0) {
      max = left;
    }

    if (right < list.size() && list.get(max).compareTo(list.get(right)) < 0) {
      max = right;
    }

    if (max != index) {
      swap(max, index);
      downheap(max);
    }
  }

  void print() {
    System.out.println(list);
  }

  int size() {
    return list.size();
  }

  T getMax() {
    return list.get(0);
  }

  public ArrayList<T> heapSort() throws Exception {
    ArrayList<T> data = new ArrayList<>();
    while (!list.isEmpty()) {
      data.add(this.remove());
    }
    return data;
  }
}
````

````java
// driver / main
Heap<Integer> h = new Heap();

System.out.println("Entered 6 keys:- 3, 10, 12, 8, 2, 14");

h.insert(3);
h.insert(10);
h.insert(12);
h.insert(8);
h.insert(2);
h.insert(14);

h.print();
System.out.println("The current size of the heap is " + h.size());
System.out.println("The current maximum element is " + h.getMax());

System.out.println("removed element : " + h.remove());
h.print();

h.insert(15);
h.print();

h.insert(5);
h.print();

System.out.println("The current size of the heap is " + h.size());
System.out.println("The current maximum element is " + h.getMax());
````

````java
PriorityQueue<Integer> minHeap = new PriorityQueue<>();

// insert
minHeap.add(50);
System.out.println("Heap: " + minHeap);
minHeap.add(10);
minHeap.add(10);
System.out.println("Heap: " + minHeap);
minHeap.add(30);
System.out.println("Heap: " + minHeap);
minHeap.add(20);
System.out.println("Heap: " + minHeap);

// read
System.out.println("Peek: " + minHeap.peek());

// remove
System.out.println("Poll: " + minHeap.poll());
System.out.println("After poll: " + minHeap);
minHeap.remove(30);
System.out.println("After removing 30: " + minHeap);

// size
System.out.println("Size: " + minHeap.size());

// empty / non empty
System.out.println("Is empty? " + minHeap.isEmpty());
````
